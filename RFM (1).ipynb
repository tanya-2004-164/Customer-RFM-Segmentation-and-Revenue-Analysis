```python

import pandas as pd

# Load the uploaded CSV file
file_path = "/content/Online_Retail_II.csv"

df = pd.read_csv(
    file_path,
    encoding="ISO-8859-1",
    on_bad_lines="warn"
)

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

display(df.head())
```

```python

# Step 1: Clean column names
df.columns = (
    df.columns
    .str.replace("ï»¿", "", regex=False)
    .str.replace("\ufeff", "", regex=False)
    .str.strip()
)

# Step 2: Verify the column names
print("Corrected columns:")
print(df.columns.tolist())

# Step 3: Check invoice values
print("\nSample invoice values:")
print(df["Invoice"].head(10).tolist())
```

```python

# Convert columns to correct data types

df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    dayfirst=True,
    errors="coerce"
)

df["Quantity"] = pd.to_numeric(
    df["Quantity"],
    errors="coerce"
)

df["Price"] = pd.to_numeric(
    df["Price"],
    errors="coerce"
)

df["Customer ID"] = pd.to_numeric(
    df["Customer ID"],
    errors="coerce"
).astype("Int64")

# Check the results
print("Updated data types:")
print(df.dtypes)

print("\nMissing values after conversion:")
print(df[["InvoiceDate", "Quantity", "Price", "Customer ID"]].isnull().sum())

display(df.head())
```

```python

# STEP 7: Identify data quality issues

# 1. Cancelled invoices usually start with C
cancelled_invoices = df["Invoice"].astype(str).str.startswith("C")

# 2. Negative quantities (often returns)
negative_quantity = df["Quantity"] < 0

# 3. Invalid or zero prices
invalid_price = df["Price"].isna() | (df["Price"] <= 0)

# 4. Missing customer IDs
missing_customer = df["Customer ID"].isna()

print("Total rows:", len(df))
print("Cancelled invoice rows:", cancelled_invoices.sum())
print("Negative quantity rows:", negative_quantity.sum())
print("Missing or non-positive price rows:", invalid_price.sum())
print("Missing Customer IDs:", missing_customer.sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python

# Keep the original df unchanged
clean_df = df.copy()

# Identify cancelled invoices
is_cancelled = clean_df["Invoice"].astype(str).str.startswith("C")

# Keep rows with valid customer, date, quantity and price
valid_rows = (
    clean_df["Customer ID"].notna()
    & clean_df["InvoiceDate"].notna()
    & clean_df["Quantity"].notna()
    & clean_df["Price"].notna()
    & (clean_df["Quantity"] > 0)
    & (clean_df["Price"] > 0)
    & (~is_cancelled)
)

clean_df = clean_df.loc[valid_rows].copy()

print("Original rows:", len(df))
print("Rows retained for purchase analysis:", len(clean_df))
print("Rows excluded:", len(df) - len(clean_df))

display(clean_df.head())
```

```python

import pandas as pd

# Reference date: one day after the last purchase date
reference_date = clean_df["InvoiceDate"].max().normalize() + pd.Timedelta(days=1)

# Calculate revenue for each transaction line
clean_df["Revenue"] = clean_df["Quantity"] * clean_df["Price"]

# Calculate RFM for each customer
rfm = clean_df.groupby("Customer ID").agg(
    Recency=("InvoiceDate", lambda x: (reference_date - x.max().normalize()).days),
    Frequency=("Invoice", "nunique"),
    Monetary=("Revenue", "sum")
).reset_index()

# Display results
print("Reference date:", reference_date.date())
print("Total customers analyzed:", len(rfm))
display(rfm.head(10))
```

```python

# Assign scores from 1 to 5 using quintiles

# Recency: lower days = better score
rfm["R_Score"] = pd.qcut(
    rfm["Recency"].rank(method="first"),
    5,
    labels=[5, 4, 3, 2, 1]
).astype(int)

# Frequency: more orders = better score
rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

# Monetary: higher spending = better score
rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

# Combine scores into one RFM score
rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str)
    + rfm["F_Score"].astype(str)
    + rfm["M_Score"].astype(str)
)

display(rfm.head(10))
```

```python

# Assign business-friendly customer segments

def assign_segment(row):
    R = row["R_Score"]
    F = row["F_Score"]
    M = row["M_Score"]

    if R >= 4 and F >= 4 and M >= 4:
        return "Champions"
    elif R >= 3 and F >= 4:
        return "Loyal Customers"
    elif R >= 4 and F <= 2:
        return "Recent Customers"
    elif R <= 2 and F >= 3:
        return "At Risk"
    elif R <= 2 and F <= 2:
        return "Hibernating"
    else:
        return "Potential Loyalists"

rfm["Segment"] = rfm.apply(assign_segment, axis=1)

# Count customers in each segment
segment_summary = (
    rfm.groupby("Segment")
    .agg(
        Customers=("Customer ID", "count"),
        Avg_Recency=("Recency", "mean"),
        Avg_Frequency=("Frequency", "mean"),
        Avg_Monetary=("Monetary", "mean"),
        Total_Revenue=("Monetary", "sum")
    )
    .round(2)
    .sort_values("Customers", ascending=False)
    .reset_index()
)

display(segment_summary)
```

```python

# Export customer-level RFM results
rfm.to_csv("RFM_Customer_Segments.csv", index=False)

# Export segment-level business summary
segment_summary.to_csv("RFM_Segment_Summary.csv", index=False)

print("Files created successfully!")
print("1. RFM_Customer_Segments.csv")
print("2. RFM_Segment_Summary.csv")
```
